```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Age":[21,22,np.nan,24,22,27,29,np.nan],
    "Study_Hours":[2.0,3.5,4.0,np.nan,3.0,5.0,6.0,4.5],
    "Department":["CSE","AIML","CSE","ECE","AIML","CSE","ECE","AIML"],
    "Passed":["Yes","Yes","No","Yes","Yes","Yes","No","Yes"]
}
df = pd.DataFrame(data)
display(df)
df.info()
print("\nMissing values:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()
for col in ["Age", "Study_Hours"]:
    df[col] = df[col].fillna(df[col].median())
for col in ["Department", "Passed"]:
    df[col] = df[col].fillna(df[col].mode()[0])

df_encoded = pd.get_dummies(df, columns=["Department"], dtype=int)
df_encoded["Passed"] = df_encoded["Passed"].map({"No":0, "Yes":1})
display(df_encoded)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_encoded[["Age_scaled","Study_Hours_scaled"]] = scaler.fit_transform(
    df_encoded[["Age","Study_Hours"]]
)
display(df_encoded)
print("Remaining missing values:\n", df_encoded.isnull().sum())
```
