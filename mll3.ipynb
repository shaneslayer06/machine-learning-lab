```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine
wine = load_wine(as_frame=True)
df = wine.frame
display(df.head())
print("Shape:", df.shape)
df.info()
display(df.describe().T)
print("Class counts:\n", df["target"].value_counts().sort_index())

plt.figure(figsize=(7,4))
sns.histplot(data=df, x="alcohol", kde=True)
plt.title("Distribution of Alcohol")
plt.show()

plt.figure(figsize=(7,4))
sns.boxplot(data=df, x="target", y="alcohol")
plt.title("Alcohol by Wine Class")
plt.show()

plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x="alcohol", y="color_intensity", hue="target")
plt.title("Alcohol vs Color Intensity")
plt.show()

cols = ["alcohol","malic_acid","ash","magnesium","color_intensity","hue"]
plt.figure(figsize=(8,6))
sns.heatmap(df[cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
